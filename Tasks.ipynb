```python
#Task1: Steps walked each day tracked on a smartwatch (line chart)

import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Running steps": [3200,5800,9100,4400,6500,12000,2100]
}

df = pd.DataFrame(data)
df.head()

```

```python

#Task4: Pokemon (scatter plot)


pokemon_data = {"Characters": ["Pikachu","Charizard", "Bulbasaur","Mewtwo", "Snorlax", "Gengar"],
                "attack": [55,84,49, 110,110,65],
                "defense": [40,78,49,90,65,60]}

df_pokemon = pd.DataFrame(pokemon_data)

color_map = {"Pikachu":"yellow", "Charizard":"orange","Bulbasaur":"green","Mewtwo":"purple","Snorlax":"blue","Gengar":"gray"}

for char, c in color_map.items():
  df_char = df_pokemon[df_pokemon["Characters"] == char]
  plt.scatter(df_char["attack"],df_char["defense"], c=c, label=char)

plt.xlabel("Attack")
plt.ylabel("Defense")
plt.title("Attack vs Defense by Characters")
plt.legend()
plt.savefig("pokemon_chart.png")
plt.show()
```
